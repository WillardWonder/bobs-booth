4. **When you're done, stop the session** (the power button at the top right) so it stops using GPU hours.

How to write for Bob:
- A blank line between paragraphs gives Bob a breath between them.
- Put a `|` before the part Bob should shout: `Here's the pitch. | It's a home run!`
- Start a paragraph with `!!` to make the whole paragraph excited.
- Wausau, NTC, .edu, AI and Josh's email are pronounced right automatically.


```python
# Warms Bob up. Run once per session.
import os, sys, urllib.request
BATCH = os.environ.get("KAGGLE_KERNEL_RUN_TYPE") == "Batch"
if BATCH:
    print("Open this notebook with Edit and Run All. A saved run does nothing, so it never spends GPU time by itself.")
else:
    os.environ["BOB_MODE"] = "voice"
    urllib.request.urlretrieve("https://raw.githubusercontent.com/WillardWonder/bobs-booth/main/kaggle/studio.py", "/kaggle/working/studio.py")
    sys.path.insert(0, "/kaggle/working")
    import studio
    studio.setup()
    studio.NUM_STEP = 32                      # best quality
    voice = studio.Voice(); voice.choose()
    writer = studio.Writer()
    from IPython.display import Audio, FileLink, display

    def bob(text, style="exact", name="bob-clip"):
        script = studio.bobify(text, writer) if style == "bob" else text
        print(script, "\n")
        mp3 = f"/kaggle/working/{name}.mp3"
        dur = studio.render_clip(voice, studio.parse_parts(script), mp3)
        print(f"{dur:.0f} seconds. Download it below, or from Output on the right (/kaggle/working/{name}.mp3).")
        display(Audio(mp3), FileLink(f"{name}.mp3"))

    print("Bob is warmed up. Edit the next cell and run it.")
```

```python
TEXT = """Well, hello again, everybody. Bob here.

They let me out of retirement one more time, just to say this. | Go Brewers!"""

STYLE = "exact"     # "exact" reads your words as written; "bob" lets Bob rewrite them in his own style first
NAME = "bob-clip"   # the MP3 file name

if not BATCH:
    bob(TEXT, STYLE, NAME)
```
